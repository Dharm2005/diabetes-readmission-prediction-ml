```python
# To reduce cardinality
# Convert ICD Codes Into Disease Groups

# This is a coarse custom grouping of selected ICD-9 diagnosis code ranges, not a complete ICD-9 classification.
# Values outside the listed ranges are grouped as "Other".

# Range       Disease Category
# ----------------------------
# 390–459     Circulatory
# 460–519     Respiratory
# 520–579     Digestive
# 580–629     Genitourinary
# 250         Diabetes
# 800–999     Injury
# 710–739     Musculoskeletal
# 140–239     Neoplasms (Cancer)
```

```python
def categorize_diagnosis(code):
  """Map diagnosis codes into the project's documented coarse groups."""
  if pd.isna(code) or str(code).strip().lower() in {
    'unknown', 'nan', 'none', ''
  }:
    return 'Unknown'

  code = str(code).strip().upper()

  if code.startswith('V'):
    return 'Supplementary'
  if code.startswith('E'):
    return 'External Injury'

  try:
    # ICD-9-CM codes may have decimals, e.g. 519.8.
    # The three-digit integer part determines the range used here.
    base_code = int(float(code))
  except (ValueError, TypeError):
    return 'Other'

  if base_code == 250:
    return 'Diabetes'
  if 140 <= base_code <= 239:
    return 'Neoplasms'
  if 390 <= base_code <= 459:
    return 'Circulatory'
  if 460 <= base_code <= 519:
    return 'Respiratory'
  if 520 <= base_code <= 579:
    return 'Digestive'
  if 580 <= base_code <= 629:
    return 'Genitourinary'
  if 710 <= base_code <= 739:
    return 'Musculoskeletal'
  if 800 <= base_code <= 999:
    return 'Injury'

  return 'Other'
```

```python
df['diag_1'] = df['diag_1'].apply(categorize_diagnosis)
df['diag_2'] = df['diag_2'].apply(categorize_diagnosis)
df['diag_3'] = df['diag_3'].apply(categorize_diagnosis)
```

```python
def apply_mapping(df, column_name, mapping_dict):
  df[column_name] = df[column_name].map(mapping_dict)
  df[column_name] = df[column_name].fillna('Unknown')
  return df
```
